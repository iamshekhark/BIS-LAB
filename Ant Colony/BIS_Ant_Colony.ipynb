```python

import numpy as np
import random

# Number of cities
NUM_CITIES = 10
NUM_ANTS = 5

# Random distance matrix
np.random.seed(42)
distance = np.random.randint(10, 100, (NUM_CITIES, NUM_CITIES))

# Make distance matrix symmetric
distance = (distance + distance.T) // 2
np.fill_diagonal(distance, 0)

# Initialize pheromone
pheromone = np.ones((NUM_CITIES, NUM_CITIES))

# ACO parameters
alpha = 1
beta = 2

# Calculate probability of selecting next city
def calculate_probability(current, unvisited):
    probabilities = []

    for city in unvisited:
        value = (pheromone[current][city] ** alpha) * \
                ((1 / distance[current][city]) ** beta)
        probabilities.append(value)

    probabilities = np.array(probabilities)
    probabilities = probabilities / probabilities.sum()

    return probabilities


# Construct route
def construct_route():
    start = random.randint(0, NUM_CITIES - 1)
    route = [start]

    unvisited = list(range(NUM_CITIES))
    unvisited.remove(start)

    while unvisited:
        current = route[-1]

        probabilities = calculate_probability(current, unvisited)

        next_city = random.choices(
            unvisited, weights=probabilities, k=1
        )[0]

        route.append(next_city)
        unvisited.remove(next_city)

    return route


# Generate routes for ants
print("ANT COLONY OPTIMIZATION")
print("-----------------------")

for ant in range(NUM_ANTS):
    route = construct_route()

    total_distance = sum(
        distance[route[i]][route[i + 1]]
        for i in range(NUM_CITIES - 1)
    )

    total_distance += distance[route[-1]][route[0]]

    print("\nAnt", ant + 1)
    print("Route:", [city + 1 for city in route])
    print("Probability-based route construction completed")
    print("Total Distance:", total_distance)

    print("All cities visited exactly once:",
          len(route) == NUM_CITIES and len(set(route)) == NUM_CITIES)
```
